```python
1. Price vs CPU_frequency

Nilai korelasi = 0.3667 (positif, sedang-lemah).

Artinya:

Semakin tinggi CPU frequency, harga cenderung lebih mahal.

Tetapi korelasi hanya sedang, bukan kuat → masih ada faktor lain yang lebih dominan menentukan harga 
(contoh: jumlah core, RAM, GPU, brand, generasi CPU).

Interpretasi: CPU_frequency memang berpengaruh terhadap harga, tapi tidak cukup kuat untuk jadi prediktor tunggal.


    
2. Price vs Screen_Size_inch

Nilai korelasi = -0.1106 (negatif lemah).

Artinya:

Ukuran layar tidak berhubungan signifikan dengan harga.

Malah ada sedikit kecenderungan laptop/PC dengan layar lebih besar harganya justru lebih murah, 
tapi ini sangat lemah dan bisa jadi efek samping data (misalnya laptop gaming 14" lebih mahal dari 17" standar).

Interpretasi: Screen size bukan faktor utama harga.


3. Price vs Weight_pounds

Nilai korelasi = -0.0503 (hampir nol).

Artinya:

Berat laptop tidak ada hubungan dengan harga.

Laptop berat tidak selalu murah atau mahal, laptop ringan juga bisa murah atau mahal → tergantung spesifikasi lain.

Interpretasi: Weight tidak relevan untuk menjelaskan variasi harga.


4. Kesimpulan

Faktor yang paling berkorelasi dengan harga adalah CPU_frequency (r = 0.3667) → makin cepat CPU, 
cenderung makin mahal.

Screen size (r = -0.1106) dan Weight (r = -0.0503) → hampir tidak ada pengaruh pada harga.

Jadi, untuk memprediksi Price, variabel yang lebih kuat selain CPU_frequency perlu ditambahkan,
misalnya: jumlah core, RAM, storage, brand, atau GPU.
```

# Task 3 - GroupBy and Pivot Tables

Group the parameters "GPU", "CPU_core" and "Price" to make a pivot table and visualize this connection using the pcolor plot.


```python

# Create the group
df_gptest = df[['GPU','CPU_core','Price']]
grouped_test1 = df_gptest.groupby(['GPU','CPU_core'],as_index=False).mean()
print(grouped_test1)
```

```python

# Create the Pivot table
grouped_pivot = grouped_test1.pivot(index='GPU',columns='CPU_core')
print(grouped_pivot)
```

```python
# Write your code below and press Shift+Enter to execute
# Create the Plot
fig, ax = plt.subplots()
im = ax.pcolor(grouped_pivot, cmap='RdBu')

#label names
row_labels = grouped_pivot.columns.levels[1]
col_labels = grouped_pivot.index

#move ticks and labels to the center
ax.set_xticks(np.arange(grouped_pivot.shape[1]) + 0.5, minor=False)
ax.set_yticks(np.arange(grouped_pivot.shape[0]) + 0.5, minor=False)

#insert labels
ax.set_xticklabels(row_labels, minor=False)
ax.set_yticklabels(col_labels, minor=False)

fig.colorbar(im)
```

# Task 4 - Pearson Correlation and p-values

Use the `scipy.stats.pearsonr()` function to evaluate the Pearson Coefficient and the p-values for each parameter tested above. This will help you determine the parameters most likely to have a strong effect on the price of the laptops.


```python
# Write your code below and press Shift+Enter to execute
for param in ['RAM_GB','CPU_frequency','Storage_GB_SSD','Screen_Size_inch','Weight_pounds','CPU_core','OS','GPU','Category']:
    pearson_coef, p_value = stats.pearsonr(df[param], df['Price'])
    print(param)
    print("The Pearson Correlation Coefficient for ",param," is", pearson_coef, " with a P-value of P =", p_value)
```
