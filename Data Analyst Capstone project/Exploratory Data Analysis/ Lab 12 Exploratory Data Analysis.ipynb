#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
missing_val = df[['Employment', 'JobSat', 'RemoteWork']].isna().sum()
print(missing_val)
```

```python
#because JobSat is float we can use median or most freq
most_freq_jobsat = df['JobSat'].mode()[0]
print(f'Most freq value in JobSat is : {most_freq_jobsat}')

median_val_jobsat = df['JobSat'].median()
print(f'Median value of JobSat is : {median_val_jobsat}')
```

```python
#Change remotework and jobsat with most req
remote_work_filled = df['RemoteWork'].fillna('Hybrid')
job_sat_filled = df['JobSat'].fillna(most_freq_jobsat) 

print(remote_work_filled) #no 4 & 5 used to be Nan now become Hybrid
print(job_sat_filled) #no 1 -4 mostly Nan now become 8.0
```

```python
df['JobSat'] = job_sat_filled
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
df[['YearsCodePro','JobSat']].head(20)
```

```python
missing_val = df[['YearsCodePro', 'JobSat']].isna().sum()
print(missing_val)
```

```python
df['YearsCodePro'].dtype
```

```python
pd.to_numeric(df["YearsCodePro"], errors="coerce")
df["YearsCodePro"] = (
    df["YearsCodePro"]
    .replace("Less than 1 year", "0.5")
    .replace("More than 50 years", "51")
    .astype(float)
)
```

```python
df[['YearsCodePro','JobSat']].head(20)
```

```python
#Create Experience Ranges
#We'll categorize developers into bins for experience levels.

bins = [0, 5, 10, 20, 100]
labels = ["0-5 years", "5-10 years", "10-20 years", ">20 years"]

df["ExperienceRange"] = pd.cut(df["YearsCodePro"], bins=bins, labels=labels, right=False)
#bins=bins agar nilai di list dapat dimasukkan 
```

```python
print(df["ExperienceRange"].value_counts())
```

```python
median_satisfaction = (
    df.groupby("ExperienceRange")["JobSat"]
      .median()
      .reset_index()
)

print(median_satisfaction)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(median_satisfaction["ExperienceRange"], median_satisfaction["JobSat"],
        color="skyblue", edgecolor="black")

plt.title("Median Job Satisfaction by Years of Professional Coding Experience")
plt.xlabel("Years of Professional Coding Experience")
plt.ylabel("Median Job Satisfaction Score")
plt.grid(axis="y", linestyle="--", alpha=0.6)

# Label bar values
for i, v in enumerate(median_satisfaction["JobSat"]):
    plt.text(i, v + 0.1, f"{v:.1f}", ha="center", fontsize=10)

plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
selected_countries = ["Indonesia", "India", "Japan"]
df_region = df[df["Country"].isin(selected_countries)]


# Split the language list and expand into rows
df_region = df_region.assign(
    Language=df_region["LanguageHaveWorkedWith"].str.split(";")
).explode("Language")

# Clean extra spaces
df_region["Language"] = df_region["Language"].str.strip()
```

```python
language_counts = (
    df_region.groupby(["Country", "Language"])
             .size()
             .reset_index(name="Count")
)
```

```python
pivot = language_counts.pivot(index="Language", columns="Country", values="Count").fillna(0)

# Keep only the top 10 languages globally
top10_langs = pivot.sum(axis=1).sort_values(ascending=False).head(10).index
pivot_top10 = pivot.loc[top10_langs]

plt.figure(figsize=(10,6))
sns.heatmap(pivot_top10, annot=True, fmt=".0f", cmap="YlGnBu")
plt.title("Popularity of Programming Languages by Country")
plt.xlabel("Country")
plt.ylabel("Programming Language")
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
#Make sure both columns are numeric (int64 or float64).
#If not, convert them:

df["YearsCodePro"] = pd.to_numeric(df["YearsCodePro"], errors="coerce")
df["JobSatPoints_1"] = pd.to_numeric(df["JobSatPoints_1"], errors="coerce")
```

```python
plt.figure(figsize=(8,6))
sns.regplot(x="YearsCodePro", y="JobSatPoints_1", data=df,
            scatter_kws={"alpha":0.5}, line_kws={"color":"red"})
plt.title("Correlation Between Years of Experience and Job Satisfaction")
plt.xlabel("Years of Professional Coding Experience")
plt.ylabel("Job Satisfaction (Points)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
#add regression line also 
```

```python
corr = df["YearsCodePro"].corr(df["JobSatPoints_1"])
print(f"Correlation between YearsCodePro and JobSatPoints_1: {corr:.2f}")
```

```python
#A positive correlation (e.g., r = 0.32) → job satisfaction tends to increase with more years of experience.

#A near-zero correlation → experience and satisfaction are mostly independent.

#A negative correlation → more experience correlates with lower satisfaction (possible burnout or stagnation).
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
#Preview your data

#Check the unique categories to understand what’s inside each column.

print(df[["EdLevel", "Employment"]].head())
#print("\nUnique Education Levels:", df["EdLevel"].unique())
#print("\nUnique Employment Types:", df["Employment"].unique())
```

```python
edu_emp_ct = pd.crosstab(df["EdLevel"], df["Employment"], normalize="index") * 100
print(edu_emp_ct.round(1))
#Cross-tabulation (crosstab) adalah teknik untuk membuat tabel yang merangkum distribusi frekuensi dari dua atau lebih variabel kategori
#pd.crosstab(df["EdLevel"], df["Employment"]): Membuat tabel 
# yang menghitung berapa banyak orang di setiap tingkat pendidikan (EdLevel) yang memiliki status pekerjaan (Employment) tertentu.
#normalize="index": Ini adalah bagian kuncinya. Alih-alih menunjukkan jumlah mentah, parameter ini mengubah angka menjadi 
# persentase per baris (indeks). Hasilnya menjawab 
# pertanyaan seperti: "Dari semua orang yang lulusan S1, berapa persen yang bekerja penuh waktu, paruh waktu, atau tidak bekerja?"
```

```python
#Visualize with a Stacked Bar Chart


edu_emp_ct.plot(kind="bar", stacked=True, colormap="Blues", figsize=(10,6))
plt.title("Employment Type Distribution by Education Level (%)")
plt.xlabel("Education Level")
plt.ylabel("Percentage of Respondents")
plt.legend(title="Employment Type", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```
