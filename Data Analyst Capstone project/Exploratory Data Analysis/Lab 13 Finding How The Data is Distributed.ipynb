### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
!pip install matplotlib-venn
```

```python
from matplotlib_venn import venn2
import matplotlib.pyplot as plt
```

```python
print(df[["LanguageHaveWorkedWith", "LanguageWantToWorkWith"]].head())
```

```python
# Split each language string into a list
df["LanguageHaveWorkedWith"] = df["LanguageHaveWorkedWith"].str.split(";")
df["LanguageWantToWorkWith"] = df["LanguageWantToWorkWith"].str.split(";")

# Remove whitespace and explode into separate rows
df_exploded_have = df.explode("LanguageHaveWorkedWith")
df_exploded_have["LanguageHaveWorkedWith"] = df_exploded_have["LanguageHaveWorkedWith"].str.strip()

df_exploded_want = df.explode("LanguageWantToWorkWith")
df_exploded_want["LanguageWantToWorkWith"] = df_exploded_want["LanguageWantToWorkWith"].str.strip()
```

```python
# Count how many respondents use each language
have_counts = (
    df_exploded_have["LanguageHaveWorkedWith"]
    .value_counts()
    .rename_axis("Language")
    .reset_index(name="HaveWorkedWith_Count")
)

# Count how many respondents want to work with each language
want_counts = (
    df_exploded_want["LanguageWantToWorkWith"]
    .value_counts()
    .rename_axis("Language")
    .reset_index(name="WantToWorkWith_Count")
)
```

```python
lang_compare = pd.merge(have_counts, want_counts, on="Language", how="outer").fillna(0)
lang_compare["HaveWorkedWith_Count"] = lang_compare["HaveWorkedWith_Count"].astype(int)
lang_compare["WantToWorkWith_Count"] = lang_compare["WantToWorkWith_Count"].astype(int)

print(lang_compare.head())
```

```python
plt.figure(figsize=(12,7))
top_langs = lang_compare.sort_values("HaveWorkedWith_Count", ascending=False).head(10)

sns.barplot(x="Language", y="HaveWorkedWith_Count", data=top_langs, color="skyblue", label="Have Worked With")
sns.barplot(x="Language", y="WantToWorkWith_Count", data=top_langs, color="orange", alpha=0.7, label="Want To Work With")

plt.title("Comparison of Programming Languages: Current vs Desired")
plt.xlabel("Programming Language")
plt.ylabel("Number of Respondents")
plt.legend()
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
have_set = set(df_exploded_have["LanguageHaveWorkedWith"].dropna().unique())
want_set = set(df_exploded_want["LanguageWantToWorkWith"].dropna().unique())

plt.figure(figsize=(6,6))
venn2([have_set, want_set],
      set_labels=("Have Worked With", "Want To Work With"),
      set_colors=("skyblue", "orange"),
      alpha=0.7)
plt.title("Overlap Between Current and Desired Programming Languages")
plt.show()
```
