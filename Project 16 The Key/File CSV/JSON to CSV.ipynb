```python
import pandas as pd
import json
import os

# --- 1. Konfigurasi File Path ---
# Path file JSON input Anda yang sebenarnya
JSON_INPUT_PATH = r"/Users/ikawahyuni/Downloads/EF_data.json"

# Nama file output Excel (akan dibuat di folder yang sama dengan skrip ini)
EXCEL_OUTPUT_PATH = 'EF_data_converted.xlsx'

# --- 2. Fungsi Konversi JSON ke XLSX ---
def json_to_xlsx(json_filepath, xlsx_filepath):
    """Mengonversi file JSON menjadi file XLSX menggunakan Pandas."""
    try:
        # Membaca file JSON
        with open(json_filepath, 'r') as f:
            data = json.load(f)

        # Menentukan format data (list atau dictionary)
        if isinstance(data, list):
            # Ini adalah skenario yang paling umum untuk data tabular
            df = pd.DataFrame(data) 
        elif isinstance(data, dict):
            # Jika data adalah dictionary (misalnya {key: [list_of_records]}), Anda mungkin perlu menyesuaikan
            # di sini kita asumsikan ini adalah satu record atau data flat
            df = pd.DataFrame([data])
        else:
            print("Struktur JSON tidak didukung untuk konversi langsung ke DataFrame.")
            return False

        # Menyimpan DataFrame ke file Excel
        # index=False agar kolom indeks (0, 1, 2, ...) tidak ikut tersimpan
        df.to_excel(xlsx_filepath, index=False)
        print(f"✅ Konversi berhasil! File XLSX tersimpan di: {xlsx_filepath}")
        return True

    except FileNotFoundError:
        print(f"❌ Error: File JSON tidak ditemukan di: {json_filepath}")
        return False
    except Exception as e:
        print(f"❌ Terjadi error saat konversi: {e}")
        return False

# --- 3. Eksekusi ---
if __name__ == '__main__':
    json_to_xlsx(JSON_INPUT_PATH, EXCEL_OUTPUT_PATH)
```

```python
import pandas as pd
import json

# 1. Tentukan nama file input dan output
INPUT_FILE = 'BC_data.json'
OUTPUT_FILE = 'BC_data_flattened.csv'

def flatten_json_to_csv(input_file, output_file):
    """
    Meratakan data JSON bersarang (nested) dari file BC_data.json
    dan mengekspor hasilnya ke file CSV.
    """
    try:
        # A. Membaca file JSON
        with open(input_file, 'r') as f:
            data = json.load(f)

        # B. Flattening Menggunakan json_normalize
        # record_path: Menunjuk ke daftar postingan (media.data)
        # meta: Mengambil detail profil (header) dan melampirkannya ke setiap postingan
        df_posts = pd.json_normalize(
            data,
            record_path=['media', 'data'],
            meta=['username', 'followers_count', 'media_count']
        )
        
        # C. Feature Engineering (Tambahan, Sesuai Pertanyaan Anda Sebelumnya)
        # Mengubah timestamp ke waktu lokal WIB dan mengekstrak fitur analisis
        df_posts['timestamp'] = pd.to_datetime(df_posts['timestamp']).dt.tz_convert('Asia/Jakarta')
        df_posts['month_posted'] = df_posts['timestamp'].dt.month_name()
        df_posts['hour_posted_wib'] = df_posts['timestamp'].dt.hour
        df_posts['caption_length'] = df_posts['caption'].str.len()
        df_posts['hashtag_count'] = df_posts['caption'].str.count('#')

        # D. Export ke CSV
        df_posts.to_csv(output_file, index=False)
        
        print(f"✅ Data berhasil diratakan dan diekspor ke: {output_file}")
        print("\n--- Kolom Hasil ---")
        print(df_posts.info())
        
        return df_posts # Mengembalikan DataFrame untuk ditampilkan

    except FileNotFoundError:
        print(f"❌ Error: File '{input_file}' tidak ditemukan.")
    except Exception as e:
        print(f"❌ Terjadi kesalahan saat memproses data: {e}")

# Jalankan fungsi utama
df_result = flatten_json_to_csv(INPUT_FILE, OUTPUT_FILE)
```
